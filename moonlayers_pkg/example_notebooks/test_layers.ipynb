# Test Layer Ordering and Vector Layers

This notebook tests:
1. Layer ordering controls
2. Vector/Feature layer support
3. Adding layers after widget is rendered

````python
from moonlayers import MoonMap
````

## Step 1: Create and Display the Map

First, create the map widget and display it. **Run this cell first and wait for the map to appear.**

````python
moon_map = MoonMap(
    projection="ESRI:103878",
    layer_switcher=True,
    controls={
        "zoom": True,
        "zoom_slider": True,
        "rotate": True,
        "scale_line": True,
        "mouse_position": {"proj": "IAU_MOON_GEOG", "precision": 4},
        "fullscreen": True
    }
)

# Display the map
moon_map
````

## Step 2: Add Layers

**After the map appears above**, run this cell to add layers.

This adds:
1. A **Feature (vector) layer** - Artemis III regions
2. Two **Mosaic (WMTS) layers** - base imagery

````python
# Add feature layer (vector polygons)
print("Adding Artemis III regions (Feature layer)...")
moon_map.add_layer('A3_Named_regions_SP')

# Add mosaic layers
print("Adding LRO WAC Mosaic...")
moon_map.add_layer('LRO_WAC_Mosaic_SPole60_100mp')

print("Adding LRO LOLA Color Shade...")
moon_map.add_layer('LRO_LOLA_ClrShade_SPole_Mosaic')

print(f"\nActive layers: {len(moon_map.active_layers)}")
for layer_id in moon_map.active_layers:
    print(f"  - {layer_id}")
````

## Step 3: Check Layer Ordering Controls

Now:
1. Open the **Layer Switcher** (top-right button)
2. Look for **↑↓ buttons** next to each Trek layer
3. Click them to reorder layers

The layer at the top of the list renders on top (most visible).

## Step 4: Search for More Layers

Find feature layers in the catalog:

````python
# Fetch all layers
layers = moon_map.fetch_trek_layers()

# Find feature layers
feature_layers = [l for l in layers 
                 if 'Feature' in l.get('serviceTypes', []) or 
                    l.get('productCat1') == 'Feature']

print(f"Found {len(feature_layers)} feature layers\n")
print("First 5 feature layers:")
for i, layer in enumerate(feature_layers[:5]):
    print(f"{i+1}. {layer['title']}")
    print(f"   productLabel: {layer['productLabel']}")
    print(f"   productCat1: {layer.get('productCat1', 'N/A')}")
    print()
````

## Step 5: Test Layer Ordering Programmatically

Check and modify layer order:

````python
# Get current state
state = moon_map.get_map_state()

print("Current layer order (bottom to top):")
for i, layer_id in enumerate(state['active_layers']):
    print(f"{i+1}. {layer_id}")

# You can reverse the order programmatically:
# reversed_order = list(reversed(state['active_layers']))
# moon_map.set_map_state({'active_layers': reversed_order})
# print("\nOrder reversed!")
````
